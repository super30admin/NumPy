Section 4 — Vectorization vs Loops

Task 4.1: Loop vs Vectorized

```python
# TODO:
# Create large array X of size 1,000,000

X = np.empty(1_000_000)
```

```python
# TODO:
# Normalize using Python loop

# HINT:
# - Time it
X = [i for i in range(1_000_000)]
start = time.time()
total = 0
for x in X:
  total += x
for i in range(len(X)):
  X[i] /= total

end = time.time()
print("time:", end-start)

```

```python
# TODO:
# Normalize using vectorization
X = np.arange(1_000_000, dtype=float)
start = time.time()
X = X/X.sum()
end = time.time()
print("time:", end-start)
```

Why is vectorization faster?
vectorization is faster because numpy executes the operation in optimized c using contiguous memory and simd, avoiding python loop overhead.


Task 4.2: Pairwise Distance (FAANG Classic)

```python
# TODO:
# Compute pairwise Euclidean distance matrix without loops

# HINT:
# - Use (x - y)^2 expansion
# - Broadcasting is key

def pairwise_distance(X):
    sq_norms = np.sum(X**2, axis=1, keepdims=True)
    dist_sq = sq_norms + sq_norms.T - 2 * X @ X.T
    return np.sqrt(dist_sq)
```

Section 5 — Numerical Stability

Task 5.1: Softmax

```python
# TODO:
# Implement naive softmax

def softmax_naive(X):
    exp_X = np.exp(X)
    return exp_X/np.sum(exp_X)
```

```python
# TODO:
# Fix numerical instability

# HINT:
# - Subtract max per row

def softmax_stable(X):
    X_shift = X - np.max(X, axis=1, keepdims=True)
    exp_X = np.exp(X_shift)
    return exp_X / np.sum(exp_X, axis=1, keepdims=True)
```

Why does subtracting max work?
because softmax is invariant to adding or subtracting the same constant form all inputs

Explain difference between dot, @, and matmul.
@ is the python operator for matrix multiplication and is equivalent to np.matmul. Both follow standard linear algebra rules and support matrix-matrix, matrix-vector, and vector-vector multiplication, with consistent behavior across dimensions.
np.dot is an older, overloaded function whose behavior depends on the input dimensions: it computes a dot product for 1d arrays, matrix multiplication for 2d arrays, and a sum-product over the last axes for highrt-dimensional arrays. For clarity ans correctness in linear algebra, @ or np.matmul is preferred over np.dot.

Section 8 — Mini Case Study

```python
# TODO:
# Given X (10000, 100):
# - Normalize features
# - Compute covariance
# - Extract top-k eigenvectors

def top_k_eigenvectors(X, k):
  mu = X.mean(axis=0, keepdims=True)
  sigma = X.std(axis=0, keepdims=True)
  sigma[sigma==0]=1.0
  Xn = (X-mu)/sigma

  n = Xn.shape[0]
  C = (Xn.T @ Xn)/(n-1)

  evals, evecs = np.linalg.eigh(C)
  idx = np.argsort(evals)[::-1]
  topk_vecs = evecs[:, idx[:k]]
  return Xn, C, topk_vecs
```

Explain each step and its ML relevance.
feature normalization rescales each column to zero mean and unit variance so models that rely on distances, variance, or gradients are well-conditioned and no single feature dominates learning. Computing the covariance matrix captures how features vary together and forms the basis of pca by summarizing second-order structure in a compact (dxd) representation. Extracting the top-k eigenvectors selects the directiosn of maximum variance, enabiling dimensionality reduction that improves generalization, reduces noise, and speeds up downstream training.


1. Where did NumPy save memory?
numby saves memory through broadcasting when subtracting the mean and dividing by the standard deviation, avoiding replication of (1xd) arrays across all rows, and by computing covariance as XTX, which yields a small (dxd) matrix rather than any large pairwise structure; additionally, using a symmetric eigen-solver(eigh) exploits matrix structure to avoid unnecessary storage and computation.

2. Where did it avoid Python overhead?
python overhead is avoided by vectorized operations(mean, std, elementwise arithmetic) that run in optimized C loops, and using BLAS/LAPACK for matrix multiplication and eigen-decomposition, eliminating per-element python loops and leveraging low-level, cache-efficient, simd-optimized routines.

3. Which operation would break at scale?
at very large feature dimensions, forming the covariance matrix and especially eigen-decomposition become infeasible due to O(d^2) memory and O(d^3) time, while at very large sample sizes the main failiure point is holding full data or normalized copy in memory, requiring streaming, randomsized svd, or incremental pca instead of full covariance and eigensolvers.

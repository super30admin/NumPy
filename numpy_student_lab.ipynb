Section 4 — Vectorization vs Loops

Task 4.1: Loop vs Vectorized

```python
# TODO:
# Create large array X of size 1,000,000

X = np.random.rand(1000000)
X
```

```python
# TODO:
# Normalize using Python loop

# HINT:
# - Time it
X_mean = np.mean(X)
X_std = np.std(X)
X_normalized_loop = np.empty_like(X)
start_time = time.time()
for i in range(X.shape[0]):
    X_normalized_loop[i] = (X[i] - X_mean) / X_std
end_time = time.time()
print(f"Time taken using loop: {end_time - start_time} seconds")
```

```python
# TODO:
# Normalize using vectorization
# Time it
X_mean = np.mean(X)
X_std = np.std(X)
start_time = time.time()
X_normalized_vectorized = (X - X_mean) / X_std
end_time_vectorized = time.time()
X_normalized_vectorized
print(f"Time taken using vectorization: {end_time_vectorized - start_time} seconds")
```

Why is vectorization faster?

- Because it utilizes SIMD (Single Instruction, Multiple data)

Task 4.2: Pairwise Distance (FAANG Classic)

```python
# TODO:
# Compute pairwise Euclidean distance matrix without loops

# HINT:
# - Use (x - y)^2 expansion
# - Broadcasting is key

def pairwise_distance(X):
    sum_X = np.sum(np.square(X), axis=1)
    D = np.sqrt(sum_X[:, np.newaxis] + sum_X[np.newaxis, :] - 2 * np.dot(X, X.T))
    return D

print("D :",pairwise_distance(np.random.rand(5, 3)))
```

Section 5 — Numerical Stability

Task 5.1: Softmax

```python
# TODO:
# Implement naive softmax

def softmax_naive(X):
    exp_X = np.exp(X)
    sum_exp_X = np.sum(exp_X, axis=1, keepdims=True)
    softmax_X = exp_X / sum_exp_X
    return softmax_X

print("softmax_naive :",softmax_naive(np.random.rand(2, 3)))

```

```python
# TODO:
# Fix numerical instability

# HINT:
# - Subtract max per row

def softmax_stable(X):
    X_max = np.max(X, axis=1, keepdims=True)
    exp_X = np.exp(X - X_max)
    sum_exp_X = np.sum(exp_X, axis=1, keepdims=True)
    softmax_X = exp_X / sum_exp_X
    return softmax_X
print("softmax_stable :",softmax_stable(np.random.rand(2, 3)))
```

Section 8 — Mini Case Study

```python
# TODO:
# Given X (10000, 100):
# - Normalize features
# - Compute covariance
# - Extract top-k eigenvectors

X = np.random.rand(10000, 100)
# Normalize features
X_mean = X.mean(axis=0, keepdims=True)
X_std = X.std(axis=0, keepdims=True)
X_normalized = (X - X_mean) / X_std
# Compute covariance matrix
cov_matrix = np.cov(X_normalized, rowvar=False)
# Extract top-k eigenvectors
eigenvalues, eigenvectors = np.linalg.eigh(cov_matrix)
k = 10  # Number of top eigenvectors to extract
top_k_indices = np.argsort(eigenvalues)[-k:]
top_k_eigenvectors = eigenvectors[:, top_k_indices]
top_k_eigenvectors
```

Explain each step and its ML relevance.

- For many operations, it views the array insted of copying.

2. Where did it avoid Python overhead?

- Numpy loops run in optimised code of C/Fortron insted of python loops

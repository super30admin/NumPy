```python
# TODO:
# Create large array X of size 1,000,000

X = np.arange(1000000)
print(X.shape)
```

```python
# TODO:
# Normalize using Python loop

# HINT:
# - Time it
Xmean = X.mean()
Xstd = X.std()
tstart = time.time()
result = np.empty_like(X)
for i in range((X.shape[0])):
   result[i] = (X[i] - Xmean) / Xstd
tend = time.time()
print(tend-tstart)


```

Task 4.2: Pairwise Distance (FAANG Classic)

```python
# TODO:
# Compute pairwise Euclidean distance matrix without loops

# HINT:
# - Use (x - y)^2 expansion
# - Broadcasting is key

def pairwise_distance(X):
    sq = np.sum(X*X,axis=1,keepdims=True)
    print(sq.shape)
    equi = sq + sq.T - 2*(X  @X.T)
    print(equi.shape)
    equi = np.maximum(equi,0.0)
    return equi

input = rng.standard_normal((100,50))
output = pairwise_distance(input)
print(output)
```

Section 5 — Numerical Stability

Task 5.1: Softmax

```python
# TODO:
# Implement naive softmax

def softmax_naive(X):
    expo = np.exp(X)
    return expo/expo.sum(axis=1,keepdims = True)
    print(expo)
    
softMaxInput = np.array([[1000,1001,1002]])
print(softMaxInput)
softmax_naive(softMaxInput)
```

```python
# TODO:
# Fix numerical instability

# HINT:
# - Subtract max per row

def softmax_stable(X):
   Z = X -X.max(axis=1,keepdims=True)
   expo = np.exp(Z)
   return expo/expo.sum(axis=1,keepdims = True)
   
softMaxStableInput = np.array([[1000,1001,1002]])
print(softMaxStableInput)
softmax_stable(softMaxStableInput)
```

Why does subtracting max work?
As we are doing the expo hence we are dividing the max value from each value.

Section 8 — Mini Case Study

```python
# TODO:
# Given X (10000, 100):
# - Normalize features
# - Compute covariance
# - Extract top-k eigenvectors
X_eig = rng.standard_normal((100000,100))
mu = X_eig.mean(axis=0,keepdims=True)
sigma = X_eig.std(axis=0, keepdims=True)
X_eign = (X_eig - mu)/sigma

Conv = (X_eign.T @ X_eign) / (X_eign.shape[0]-1)

eigvals, eigvecs = np.linalg.eigh(Conv)
k=5
top_idx = np.argsort(eigvals)[-k:][::-1]
top_idx

```

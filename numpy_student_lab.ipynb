Section 4 — Vectorization vs Loops

Task 4.1: Loop vs Vectorized

```python
# TODO:
# Create large array X of size 1,000,000
range = np.random.default_rng(2)
X = range.standard_normal(1_000_000)
print(X.shape)
```

```python
# TODO:
# Normalize using Python loop

# HINT:
# - Time it
import builtins
mu = X.mean()
sigma = X.std()
t0 = time.time()
out_loop = np.empty_like(X)
for i in builtins.range(X.shape[0]):
  out_loop[i] = (X[i] - mu) / sigma
time_taken = time.time() - t0
print(round(time_taken, 4))
```

```python
# TODO:
# Normalize using vectorization
t1 = time.time()
out_vec = (X - mu)/sigma
vec_time = time.time() - t1
round(vec_time, 4)
print(round(time_taken/max(vec_time, 1e-12), 1))
print('allclose :', np.allclose(out_loop, out_vec))
```

Why is vectorization faster?

Answer: Because for loops the datatypes is left for system to decide and also we have to iterate through all the elements, while vectorization performs operations parallely.

Task 4.2: Pairwise Distance (FAANG Classic)

```python
# TODO:
# Compute pairwise Euclidean distance matrix without loops

# HINT:
# - Use (x - y)^2 expansion
# - Broadcasting is key

def pairwise_distance(X):
  sq = np.sum(X * X, axis = 1, keepdims = True)
  dist = sq + sq.T - 2 * (X @ X.T)
  dist = np.maximum(dist, 0.0)
  return np.sqrt(dist)

rng = np.random.default_rng(3)
Xsmall = rng.standard_normal((200, 10))
D = pairwise_distance(Xsmall)
print(D.shape)
```

Section 5 — Numerical Stability

Task 5.1: Softmax

```python
# TODO:
# Implement naive softmax

def softmax_naive(X):
  ex = np.exp(X)
  print(ex)
  return ex/ex.sum(axis = 1, keepdims = True)

Xbig = np.array([[1000.0, 1001.0, 1002.0]])
try:
  softmax_naive(Xbig)
except FloatingPointError as e:
  print('overflow :', e)
```

```python
# TODO:
# Fix numerical instability

# HINT:
# - Subtract max per row

def softmax_stable(X):
    Z = X - X.max(axis = 1, keepdims = True)
    ex = np.exp(Z)
    return ex/ex.sum(axis = 1, keepdims = True)

print('stable :', softmax_stable(Xbig))
print('rowSum :', softmax_stable(Xbig).sum(axis = 1))
```

Section 8 — Mini Case Study

```python
# TODO:
# Given X (10000, 100):
# - Normalize features
# - Compute covariance
# - Extract top-k eigenvectors
range = np.random.default_rng(8)
X = range.standard_normal((10_000, 100))
# Normalize
mu = X.mean(axis = 0, keepdims = True)
sigma = X.std(axis = 0, keepdims = True)
Xn = (X - mu)/sigma
# Covariance
Covariance = (Xn.T @ Xn) / (Xn.shape[0] - 1)
# Top-k Eigenvectors
eigvals, eigvecs = np.linalg.eigh(Covariance)
k = 5
top_idx = np.argsort(eigvals)[-k:][::-1]
top_idx
```

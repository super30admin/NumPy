Section 3 — Broadcasting

Task 3.1: Broadcasting Rules

```python
# TODO:
# Create A (1000, 50) and b (50,)

A = np.arange(50*1000).reshape(1000,50)
b = np.arange(50)
```

```python
# TODO:
# Add b to each row of A

# HINT:
# - No reshape required

C = A + b
print (C.shape)

# Broadcasting helps perform operations between arrays of different sizes
# IT starts by trying to match sizes from right most dimension for each of the arrays
# if the size is a match, then it moves to the left.It keeps moving to the left until at least one of the arrays runs out of dimensions, and from then on, the array with lower dimensiosn - it gets added with as many dimensions as the bigger array has (each missing dimension with size same as the other array's size for that dimension)
# If the size is not a match, it checks if one of them is has size 1. If it does, then it expands that one to the size of the other one
# If neither of them are size 1, then we get size mismatch error
```

```python
# TODO:
# Normalize each row of A

# HINT:
# - Axis matters
# - Keep dimensions in mind

X = np.random.rand(10,4)

Y = (X- np.mean(X,axis=1,keepdims=True))/np.std(X,axis=1,keepdims=True)

print (X.shape)
print (Y.shape)

print (X)
print (Y)
```

Section 4 — Vectorization vs Loops

Task 4.1: Loop vs Vectorized

```python
# TODO:
# Create large array X of size 1,000,000

rng = np.random.default_rng(2)
X = rng.standard_normal(1000000)
```

```python
# TODO:
# Normalize using Python loop

# HINT:
# - Time it

import time

mean = X.mean()
sigma = X.std()

t0 = time.time()
output = np.empty_like(X)

for i in range(X.shape[0]) :
  output[i] = (X[i]-mean)/sigma

print("Time taken = ", time.time()-t0)

```

```python
# TODO:
# Normalize using vectorization

t1 = time.time()
output_vec = (X - mean)/sigma
print ("Time taken = ", time.time()-t1)

print (np.allclose(output, output_vec))

#Vectorization is faster due to contiguous memory allocation and hence faster access
```

Why is vectorization faster?

Section 5 — Numerical Stability

Task 5.1: Softmax

```python
# TODO:
# Implement naive softmax

def softmax_naive(X):
    return np.exp(X) / np.sum(np.exp(X))
```

```python
# TODO:
# Fix numerical instability

# HINT:
# - Subtract max per row

def softmax_stable(X):
    max_row = np.max(X,axis=1, keepdims=True)
    return np.exp(X-max_row) / np.sum(np.exp(X)-max_row)


#Subtracting max works because it ensures non of the values are very large (e ^ value) can get very large, but if we substract e ^ (value-max_value) ..then the number does not get too big
```

```python
array = np.random.rand(10,10)

softmax_naive_vec = softmax_naive(array)
softmax_stable_vec = softmax_stable(array)

print (softmax_naive_vec)
print (softmax_stable_vec)
```

Why does subtracting max work?

Section 8 — Mini Case Study

```python
# TODO:
# Given X (10000, 100):
# - Normalize features
# - Compute covariance
# - Extract top-k eigenvectors

X = np.random.rand(10000,100)

#X = (X - np.mean(X))/ np.std(X)

mean = np.mean(X, axis=0,keepdims=True)
std = np.std(X, axis=0,keepdims=True)


X = (X-mean) / std

cov = (X.T @ X) /(X.shape[0] - 1)
eig_vals, eig_vecs = np.linalg.eigh(cov)

k = 5

top_indices = np.argsort(eig_vals)[-k: ][::-1]
print (top_indices)

top_eigenvectors = eig_vecs[top_indices]
print (top_eigenvectors.shape)
#eigen = np.eigenvectors(X)
#print (cov, eigen)

# First we are normalizing the features across all rows.
# Then we compute the variance matrix
# Eigen decomposition of this co-variance matrix is telling us which direction has the highest variance
# Selecting the top-k eigen vectors basically gives us the reduced diension representation of the most important dimensions of the data represented in a different space
# This is basically achieving PCA
```
